# Knowledge Discovery Frameworks: KDD, CRISP-DM, SEMMA

* **KDD (Knowledge Discovery in Databases)** -- the original academic framework (Fayyad, Piatetsky-Shapiro & Smyth, 1996). Five stages: **Selection -> Preprocessing -> Transformation -> Data Mining -> Interpretation/Evaluation**. "Knowledge discovery" as a phrase comes directly from this framework.
* **CRISP-DM (Cross-Industry Standard Process for Data Mining)** -- the framework most widely used in industry today. Six stages, business-first: **Business Understanding -> Data Understanding -> Data Preparation -> Modeling -> Evaluation -> Deployment**.
* **SEMMA (Sample, Explore, Modify, Model, Assess)** -- SAS's own branded version of essentially the same pipeline. Mostly a historical reference now, tied closely to the SAS software ecosystem.
* **TDSP (Team Data Science Process)** -- Microsoft's own methodology, launched in 2016. Rather than replacing KDD/CRISP-DM, Microsoft built TDSP as an agile layer on top of them -- essentially CRISP-DM combined with Scrum, adding team roles, project structure, and tooling (Azure-oriented) for organizations standardizing across multiple teams. Microsoft explicitly designed it to be usable *alongside* an existing lifecycle like CRISP-DM or KDD, not instead of one.
* All four describe the *same underlying logic* with different emphasis: KDD is discovery-focused and academic, CRISP-DM is project-management-focused and industry-standard, SEMMA is tool-specific, and TDSP is a team/tooling layer built on top of the others.
* None of these frameworks anticipated embeddings, knowledge graphs, or foundation models -- but the pipeline logic (understand data -> prepare it -> model it -> ask whether the result is real) still holds underneath modern AI/data science workflows.


### A side-by-side view of the four pipelines

```python
import matplotlib.pyplot as plt

frameworks = {
    "KDD (1996)": ["Selection", "Preprocessing", "Transformation", "Data Mining", "Interpretation\n& Evaluation"],
    "CRISP-DM": ["Business\nUnderstanding", "Data\nUnderstanding", "Data\nPreparation", "Modeling", "Evaluation", "Deployment"],
    "SEMMA": ["Sample", "Explore", "Modify", "Model", "Assess"],
    "TDSP (Microsoft)": ["Business\nUnderstanding", "Data Acquisition\n& Understanding", "Modeling", "Deployment", "Customer\nAcceptance"],
}

fig, axes = plt.subplots(len(frameworks), 1, figsize=(11, 8))

for ax, (name, stages) in zip(axes, frameworks.items()):
    for i, stage in enumerate(stages):
        ax.add_patch(plt.Rectangle((i * 2, 0), 1.7, 1, edgecolor='black', facecolor='#dbeafe'))
        ax.text(i * 2 + 0.85, 0.5, stage, ha='center', va='center', fontsize=8)
        if i < len(stages) - 1:
            ax.annotate('', xy=(i * 2 + 2, 0.5), xytext=(i * 2 + 1.7, 0.5),
                         arrowprops=dict(arrowstyle='->', lw=1.2))
    ax.set_xlim(-0.3, 6 * 2)
    ax.set_ylim(-0.3, 1.3)
    ax.axis('off')
    ax.set_title(name, fontsize=10, loc='left')

plt.tight_layout()
plt.show()
```

### Quick comparison

| | KDD | CRISP-DM | SEMMA | TDSP |
|---|---|---|---|---|
| Origin | Academic (1996) | Industry consortium | SAS Institute | Microsoft (2016) |
| Focus | Discovery process | Business project lifecycle | Modeling workflow | Team collaboration + tooling |
| Common use today | Conceptual reference | Most-used industry standard | Mostly historical / SAS-specific | Azure-oriented enterprise teams |
| Includes deployment? | Not explicitly | Yes | No | Yes |
| Relationship to the others | Origin of the term "knowledge discovery" | Incorporates ideas from KDD and SEMMA | Parallel, SAS-specific take | Built as a layer that can run *alongside* CRISP-DM or KDD, not replace them |
